# Data Types in Python

### A data type is a classification of data (such as string, integer, list, set, and dictionary)
### stored in a variable in memory, which determines the different operations that can be performed on it.

# Python Built-in Data Types

Python has the following built-in data types, which are categorized into the following groups:

| Category | Data Types |
|---|---|
| **Text Type** | `str` |
| **Numeric Types** | `int`, `float`, `complex` |
| **Sequence Types** | `list`, `tuple`, `range` |
| **Mapping Type** | `dict` |
| **Set Types** | `set`, `frozenset` |
| **Boolean Type** | `bool` |
| **Binary Types** | `bytes`, `bytearray`, `memoryview` |
| **None Type** | `NoneType` |

# 1. TEXT TYPE : STRING (STR)

```python
name = "Prajwal Patil"
course = "Python Course"
print(name)
print(course)

print("The type of variable name is",type(name))
print("The type of variable course is",type(course))
```

# 2. NUMERIC TYPE : INT , FLOAT , COMPLEX 

```python
# ==========================================
#             INTEGER (int)
# ==========================================

age = 25
students = 50
temperature = 45
kg = 500
days = 30

print("*" * 15 + " " + "INT" + " " + "*" * 20)
print("My age is:", age)
print("There are", students, "students in the class")
print("Today in our city the temperature is:", temperature, "°C")
print(f"I have bought {kg} kg of potatoes")
print("There are", days, "days in this month")


# ==========================================
#             FLOAT (float)
# ==========================================

celsius = 36.5
price = 499.99
height = 5.8
weight = 65.5
distance = 12.75

print("\n" + "*" * 15 + " " + "FLOAT" + " " + "*" * 20)
print("My body temperature is:", celsius, "°C")
print(f"The price of the product is: ₹{price}")
print("My height is:", height, "feet")
print("My weight is:", weight, "kg")
print("The distance to the office is:", distance, "km")


# ==========================================
#             COMPLEX (complex)
# ==========================================

voltage = 3 + 4j
current = 2 + 5j
signal = 10 + 3j
number = 7 + 2j

print("\n" + "*" * 15 + " " + "COMPLEX" + " " + "*" * 20)
print("Voltage represented as a complex number:", voltage)
print("Current represented as a complex number:", current)
print("Signal represented as a complex number:", signal)
print("Another complex number:", number)

print("Real part of voltage:", voltage.real)
print("Imaginary part of voltage:", voltage.imag)
```

# 3. SEQUENCE TYPE : LIST , TUPE , RANGE  

```python
# -------------------- LIST --------------------
# Example: Shopping items

shopping_items = ["Milk", "Bread", "Eggs", "Rice"]

print("*" * 15 + " " + "LIST" + " " + "*" * 20)
print("Shopping items:", shopping_items)
print(type(shopping_items))

# -------------------- TUPLE --------------------
# Example: Employee information

employee = ("Rahul", 25, "Developer", 50000)

print("\n" + "*" * 15 + " " + "TUPLE" + " " + "*" * 20)
print("Employee information:", employee)
print(type(employee))

# -------------------- RANGE --------------------
# Example: Student roll numbers

roll_numbers = range(101, 106)

print("\n" + "*" * 15 + " " + "RANGE" + " " + "*" * 20)
print("Student roll numbers:", list(roll_numbers))

# Example: Even numbers
even_numbers = range(2, 11, 2)

print("Even numbers:", list(even_numbers))
print(type(roll_numbers))
```

# 4. MAPPING TYPE : DICT

```python
# ============================================================
# 4. MAPPING TYPE : DICT
# ============================================================

student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "marks": 85
}

employee = {
    "id": 101,
    "name": "Priya",
    "department": "IT",
    "salary": 55000
}

product = {
    "name": "Laptop",
    "price": 65000,
    "brand": "Dell"
}

print("*" * 15 + " DICT " + "*" * 20)

print("Student:", student)
print("Student name:", student["name"])

print("Employee:", employee)
print("Employee department:", employee["department"])

print("Product:", product)
print("Product price:", product["price"])

print("\nData type:", type(student))
```

# 5. Set Types: set, frozenset

```python
# SET - Mutable collection of unique values
cities = {"Bangalore", "Mumbai", "Delhi", "Mumbai"}

print("*" * 15 + " SET " + "*" * 20)
print("Cities:", cities)
print("Data type:", type(cities))

# Adding a new value
cities.add("Chennai")
print("Updated cities:", cities)


# FROZENSET - Immutable set
skills = frozenset(["Python", "SQL", "Excel", "Python"])

print("\n" + "*" * 15 + " FROZENSET " + "*" * 20)
print("Skills:", skills)
print("Data type:", type(skills))
```

# 6. Boolean Type: bool

```python
is_logged_in = True
is_payment_completed = False
is_student = True
has_driving_license = False

print("*" * 15 + " BOOL " + "*" * 20)

print("User logged in:", is_logged_in)
print("Payment completed:", is_payment_completed)
print("Is student:", is_student)
print("Has driving license:", has_driving_license)

print("\nData type:", type(is_logged_in))


# Boolean values are commonly used in conditions
age = 20

print("\nIs the person an adult?", age >= 18)
print("Data type:", type(age >= 18))
```

# 7. Binary Types	bytes, bytearray, memoryview &NONETYPE

```python
# BYTES - Immutable sequence of bytes
data = b"Hello Python"

print("*" * 15 + " BYTES " + "*" * 20)
print("Data:", data)
print("Data type:", type(data))


# BYTEARRAY - Mutable sequence of bytes
file_data = bytearray(b"Python")

print("\n" + "*" * 15 + " BYTEARRAY " + "*" * 20)
print("Before modification:", file_data)

file_data[0] = 74

print("After modification:", file_data)
print("Data type:", type(file_data))


# MEMORYVIEW - View of binary data without copying it
numbers = bytearray(b"ABC")
view = memoryview(numbers)

print("\n" + "*" * 15 + " MEMORYVIEW " + "*" * 20)
print("Memory view:", view)
print("First byte:", view[0])
print("Data type:", type(view))


# NONE TYPE
result = None

print("\n" + "*" * 15 + " NONE " + "*" * 20)
print("Result:", result)
print("Data type:", type(result))
```
